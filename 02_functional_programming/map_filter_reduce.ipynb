```python
from functools import reduce
```

```python
data = [10, 20, 30, 40, 50, 60, 70, 80, 90]
```

#### Como funciona a lógica `filter` → `map` → `reduce`.
- `filter`: Quando temos uma lista de elementos podemos não querer trabalhar com todos, o `filter` entra aqui;
- `map`: Após termos os elementos com que queremos trabalhar podemos aplicar algum tratamento a eles, o `map` é quem faz isso;
- `reduce`: Depois de filtrar e aplicar alguma função, podemos utilizar o `reduce` que aplica alguma função de agrupamento em nossos dados.



#### Filter

```python
# no filter a função deve produzir True ou False
filter_ = filter(lambda x: x % 3 == 0, data)
print(filter_)
```

#### Map

```python
# o map aplica uma transformação nos dados
map_ = map(lambda x: x + 1, filter_)
print(map_)
```

```python
# pegamos os valores
result = list(map_)
print(result)
```

#### Reduce

```python
# no reduze "agregamos" os dados
filter_ = filter(lambda x: x % 3 == 0, data)
map_ = map(lambda x: x + 1, filter_)
reduce(lambda x, y: x + y, map_)
```

```python
# vamos:
# (filter): filtrar números maiores que 30
# (map) multiplicar por 2
# (reduce) somar tudo
result = reduce(
    lambda x, y: x + y,
    map(
        lambda x: x * 2,
        filter(lambda x: x > 30, data)
    )
)
print(result)
```
