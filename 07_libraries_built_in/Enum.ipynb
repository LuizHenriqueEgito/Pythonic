```python
from enum import (
    Enum,
    auto,
    IntEnum,
    StrEnum,
    Flag,
    unique
)
```

`Enum` cria conjuntos de valores nomeados, isso **padroniza** opções.

```python
class Status(Enum):
    PENDENTE = 1
    APROVADO = 2
    REPROVADO = 3

status = Status.APROVADO
```

```python
print(f'{status=}')
print(f'{status.name=}')
print(f'{status.value=}')
```

Podemos usar `auto()` quando o valor númerico não importa.

```python
class Color(Enum):
    RED = auto()
    GREEN = auto()
    BLUE = auto()
```

```python
color = Color.RED
```

```python
print(f'{color=}')
print(f'{color.name=}')
print(f'{color.value=}')
```

Usamos `IntEnum` quando queremos que esse nosso conjunto de opções se comporte como um numero e `StrEnum` quando queremos que se comportem como string.

```python
class Prioridade(IntEnum):
    ALTA = 3
    MEDIA = 2
    BAIXA = 1
```

```python
print(Prioridade.ALTA > Prioridade.BAIXA)
print(Prioridade.MEDIA + 1)
```

```python
class Status(StrEnum):
    APROVADO = "aprovado"
    REPROVADO = "reprovado"

status = Status.APROVADO

print(status)
print(status == "aprovado")
print(f"{status}")
print(isinstance(status, str))

status.upper()
```

Com `Flag` podemos combinar opções.

```python
class Permissao(Flag):
    LER = auto()
    ESCREVER = auto()
    EXECUTAR = auto()
```

```python
permission = Permissao.LER | Permissao.ESCREVER
```

```python
permission
```

```python
Permissao.LER in permission
```

`@unique` evita valores duplicados.

```python
@unique
class Status(Enum):
    PENDENTE = 1
    APROVADO = 2
    REPROVADO = 2  # erro
```

```python
@unique
class Status(Enum):
    PENDENTE = 1
    APROVADO = 2
    REPROVADO = 3
```

Podemos criar `methods` dentro do nosso `enum`. 

```python
@unique
class Status(Enum):
    PENDENTE = 1
    APROVADO = 2
    REPROVADO = 3

    def apporoved(self):
        return self == Status.APROVADO
```

```python
status = Status.REPROVADO
```

```python
print(status.apporoved())
```
